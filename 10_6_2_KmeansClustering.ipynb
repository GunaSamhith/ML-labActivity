```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
pixels = pd.read_csv(
    "Image_Pixel_Dataset_396x396x3.csv"
)

pixels.head()
```

```python
image_array = pixels.values

print(image_array.shape)
```

```python
kmeans = KMeans(
    n_clusters=16,
    random_state=42
)

kmeans.fit(image_array)
```

```python
compressed_pixels = kmeans.cluster_centers_[kmeans.labels_]

compressed_pixels = compressed_pixels.astype(np.uint8)
```

```python
compressed_image = compressed_pixels.reshape(396,396,3)

compressed_image.shape
```

```python
plt.figure(figsize=(6,6))

plt.imshow(compressed_image)

plt.axis("off")

plt.title("Image Compression using K-Means")

plt.show()
```
