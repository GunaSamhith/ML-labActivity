```python
import pandas as pd

data = pd.DataFrame({
    'Brightness': [40, 50, 60, 10, 70, 60, 25],
    'Saturation': [20, 50, 90, 25, 70, 10, 80],
    'Class': ['Red', 'Blue', 'Blue', 'Red',
              'Blue', 'Red', 'Blue']
})

display(data)
```

```python
new_brightness = 45
new_saturation = 40
```

```python
import math

data['Distance'] = data.apply(
    lambda row: math.sqrt(
        (row['Brightness'] - new_brightness) ** 2 +
        (row['Saturation'] - new_saturation) ** 2
    ),
    axis=1
)

print("Distances from new point:")
display(data.sort_values('Distance'))
```

```python
k = 3

nearest = data.sort_values(
    'Distance'
).head(k)

print("Nearest Neighbors:")
display(nearest)
```

```python
prediction = nearest['Class'].mode()[0]

print("New Entry:")
print("Brightness:", new_brightness)
print("Saturation:", new_saturation)

print("\nPredicted Class:", prediction)
```

```python
from sklearn.neighbors import KNeighborsClassifier

X = data[['Brightness', 'Saturation']]
y = data['Class']

model = KNeighborsClassifier(
    n_neighbors=3,
    metric='minkowski',
    p=2
)

model.fit(X, y)

new_point = [[45, 40]]

prediction = model.predict(new_point)

print("Predicted Class:", prediction[0])
```

```python
new_point = pd.DataFrame({
    'Brightness': [45],
    'Saturation': [40]
})

prediction = model.predict(new_point)

print("Predicted Class:", prediction[0])
```
